```python
import numpy as np
import pandas as pd
import glob
import re
import nltk
#from nltk.tokenize import word_tokenize
#nltk.download("punkt")
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from collections import defaultdict
from tqdm import tqdm
#from spellchecker import SpellChecker ausgesetzt, da der Prozess zu langsam wird


#Alle csv- Dateien aus der angegebenen Quelle werden verarbeitet
dataset_path = glob.glob("C:\\Users\\Benedikt Thissen\\OneDrive\\Dokumente\\1Studium\\Data Analysis\\Datensettest\\*.csv")

#Die Werte der csvs werden in tweets gelistet
tweets = []
for data in tqdm(dataset_path, desc="Verarbeite CSV-Dateien"):
    tw = pd.read_csv(data)
    tweets.append(tw)
    
#?
all_tweets = pd.concat(tweets, ignore_index=True)

#Für den Corpus werden alle Einträge aus der Spalte content verwendet
corpus = all_tweets["content"].dropna().tolist()

#Textbereinigung (V2 vocabulary auf *set* geändert, damit keine doppelte Einträge vorhanden sind)
reviews = []
vocabulary = set()
sentences = []
stop_words = set(stopwords.words("english"))
lemmatizer = WordNetLemmatizer()
#spell = SpellChecker()

#Fortschrittsanzeige der Textbereinigung
for r in tqdm(corpus, desc="Verarbeite Tweets"):
    tokens = re.findall(r'\b\w+\b', r.lower())
    #tokens = [spell.correction(word) for word in tokens_unchecked]
    rev = [lemmatizer.lemmatize(w) for w in tokens if w and w.isalpha() and w not in stop_words]
    sentences.append(rev)
    vocabulary.update(rev)

len_vocabulary = len(vocabulary)
print(f"Anzahl Wörter: {len_vocabulary}")

index = {w: i for i, w in enumerate(sorted(vocabulary))}
keys = list(index.keys())
for i in range(10):
    word = keys[i+10]
    print(f"{index[word]}|{word}")
```
